## 一、 問題說明與分治核心
本作業依據《Lecture #7 分治法》講義規範，使用分治法 (Divide and Conquer) 解決經典的河內塔問題。目標是將 $N$ 個盤子由 $A$ 柱（起點）移動至 $C$ 柱（目標），並以 $B$ 柱作為中繼站。根據講義第 64 頁所述，河內塔的核心分治邏輯可拆解為以下三個標準步驟：


1. 【拆解】 將上方 $n-1$ 個盤子由 $A$ 柱移到 $B$ 柱（清空上層空間）。
2. 【解決 Base Case】 將最底層第 $n$ 個盤子由 $A$ 柱移到 $C$ 柱。
3. 【合併】 將 $B$ 柱上的 $n-1$ 個盤子移到 $C$ 柱（完成整體搬移）。


此邏輯正對應講義中「清空上層空間，使底層盤子可以輕鬆移動」的工程語義，透過解決每個小的 Base Case 來逐步構築整體問題的解答。

## 二、 Python 程式碼實作：
遞迴版河內塔
依據作業規範與講義實作結構，下達指令請 AI 撰寫包含移動計數器的河內塔分治法遞迴程式：

```python
def hanoi_recursive(n, source, auxiliary, target, counter):
    """
    使用分治法（Divide and Conquer）解決河內塔問題並統計移動次數
    n: 當前盤數
    source: 起點柱 (pillar_from)
    auxiliary: 中繼柱 (pillar_mid)
    target: 目標柱 (pillar_to)
    counter: 用於傳遞與記錄總移動次數的 list
    """
    # Base Case: 對應講義第 64 頁實作邊界
    if n == 0:
        return

    # 步驟 1: 將 n-1 個盤子由起點移到中繼站 (講義 Step 1)
    hanoi_recursive(n - 1, source, target, auxiliary, counter)

    # 步驟 2: 移動最底層的目標盤子 (講義 Step 2)
    counter[0] += 1

    # 步驟 3: 將中繼站的 n-1 個盤子移到目的地 (講義 Step 3)
    hanoi_recursive(n - 1, auxiliary, source, target, counter)

# 實驗環境模擬
if __name__ == "__main__":
    N = 30
    counter = [0]

    # 由於 N=30 實際執行會耗費較長時間，此處同時以數學公式驗證
    moves_formula = 2 ** N - 1

    print(f"--- N = {N} 實驗結果 ---")
    print(f"理論總移動次數 (2^N - 1) = {moves_formula} 次")
```

$N = 30$ 執行結果與分析
河內塔移動次數公式為：

$T(N)=2^N-1$

當 $N = 30$ 時，系統理論輸出為：
$$2^{30} - 1 = 1,073,741,823 \text{ 次}$$
評估分析：
正如同作業解答範例所述，當 $N=30$ 時，總共需要進行約 10.7 億次移動。在現實計算機中，即便我們極致優化程式且不列印 I/O，假設每次移動與計數只需 1 微秒 ($\mu s = 10^{-6}$ 秒)，整體運行時間仍需要：
$$\frac{1,073,741,823 \times 10^{-6} \text{ 秒}}{60} \approx 17.89 \text{ 分鐘}$$
這充分證明了分治法雖然在程式撰寫上極其精簡，但若子問題的數量隨規模呈指數增長，在時間成本上便會迅速迎來組合爆炸。

## 三、 規模增長時的組合爆炸（$N=64$ 物理極限）
河內塔移動次數隨規模變化的數據如下表所示：

N = 64 的物理極限

當 N = 64：

$
2^{64}-1 = 18,446,744,073,709,551,615
$

若每次移動需 1 微秒：

$
\frac{2^{64}-1}{10^6 \times 60 \times 60 \times 24 \times 365}

$

若假設每秒可處理 1 億次移動：

$
\frac{2^{64}-1}{10^8 \times 60 \times 60 \times 24 \times 365}
\approx 584,5 \text{ 年}
$

因此，河內塔在 N=64 時，不只是「程式寫得出來但跑很久」，而是已經碰到現實物理世界的時間尺度極限。

結論： > 58 萬年這個時間跨度已遠超人類文明的歷史。$N=64$，在現實物理世界中是絕對無法完成的。瓶頸不單是「程式演算法的邏輯能不能寫出來」或是「硬體記憶體夠不夠大」，而是問題本身的動作數量（次數空間）已經遠遠跨越了現實的時間尺度。

```python
moves_64 = 2 ** 64 - 1
seconds = moves_64 / 1_000_000  # 假設單次移動僅需 1 微秒
years = seconds / (60 * 60 * 24 * 365.25)
print(f"N=64 所需時間: {years:.2f} 年")
```

##四、極限挑戰：$N = 1,000,000$ 的遞迴深度與系統崩潰測試
若在遞迴版本中強行執行 $N = 1,000,000$ 的深度測試：

```python
def hanoi_depth_test(n):
    if n == 0:
        return
    hanoi_depth_test(n - 1)
    hanoi_depth_test(n - 1)

try:
    hanoi_depth_test(1_000_000)
except RecursionError as e:
    print(f"系統捕獲異常 -> RecursionError: {e}")
```

#### 遞迴深度與系統崩潰（Stack Overflow）原因分析
Python 系統預設的遞迴深度限制（通常在 1000 層左右）。當 $N = 1,000,000$ 時，意味著演算法必須往下展開極深的 **呼叫堆疊（Call Stack）**。在微觀的記憶體架構中，每一次函式自己呼叫自己，作業系統就必須在記憶體棧區（Stack Area）開闢一個 **「棧幀（Stack Frame）」**，用來封存當前層級的區域變數、參數與程式執行中斷點。當這 100 萬個棧幀像磚塊一樣不斷向上堆疊，而底層卻遲遲無法遇到 Base Case 進行 return 釋放時，有限的棧區記憶體空間會立刻被擠爆，進而引發硬體層面的 **Stack Overflow**。Python 則會透過彈出 RecursionError 來觸發系統自我保護。

## 五、 架構優化：非遞迴 Iterative Stack（迭代版）替代方案
為了突破系統 Call Stack 的實體限制，工程實務上會使用自訂的資料結構（如明確的 List 作為 Stack）來模擬作業系統引導分治法的行為。以下為非遞迴版本的 Python 實作：

```python
def hanoi_iterative(n, source, auxiliary, target):
    """
    使用自訂 Stack 模擬遞迴的分治行為 (Iterative 替代方案)
    """
    # 堆疊內儲存任務狀態：(當前盤數, 起點, 中繼, 終點, 是否已準備好移動底盤)
    stack = [(n, source, auxiliary, target, False)]
    move_count = 0

    while stack:
        n, src, aux, tgt, ready_to_move = stack.pop()

        if n == 0:
            continue

        if ready_to_move:
            # 相當於遞迴版本中 step 2 的實質移動與計數行為
            move_count += 1
        else:
            # 依據 Stack「後進先出 (LIFO)」特性，將步驟反向壓入堆疊：
            # 3. 將中繼站的 n-1 個盤子移到目的地
            stack.append((n - 1, aux, src, tgt, False))
            # 2. 移動當前的底盤 (標記為 True，下次 pop 時直接計數)
            stack.append((1, src, aux, tgt, True))
            # 1. 將 n-1 個盤子從起點移到中繼站
            stack.append((n - 1, src, tgt, aux, False))

    return move_count

if __name__ == "__main__":
    # 理論測試 N = 30 (此處用以驗證邏輯等效性)
    # 注意：由於此題本身具備 2^N 組合爆炸特性，大數值仍需耗費計算時間
    print("迭代版 N=30 計算所得搬移次數:", hanoi_iterative(30, "A", "B", "C"))
```

### 迭代版與遞迴版的對照比較

* **遞迴版本**：仰賴作業系統底層自動管理 Call Stack。程式邏輯最為直覺，但極易受到系統環境限制而引發記憶體溢位。
* **迭代版本**：改用自訂的 Data Stack。因為其數據儲存於電腦作業系統的**堆疊區（Heap Area）**，其容納上限取決於電腦整體的實體記憶體（RAM）。在面對超大規模工程計算時（例如大型結構點雲、海量有限元素矩陣），將遞迴改寫為迭代是防止系統崩潰的必要優化手段。


## 六、關鍵核心：Base Case 的不可或缺性
在分治演算法中，最重要的一行防禦線即是：

```python
if n == 0: return
```

這就是分治法的 **基底案例 (Base Case)**。

### 批判論述
Base Case 是分治演算法中絕對的 「Return 點與終止邊界」。若在實作時遺漏了此行條件，大事化小的過程將失去「康復出口」。

當程式呼叫到 n = 0 時，它不會停止，而是盲目地繼續切分子問題，朝著 n = -1、n = -2 永無止境地遞迴下去。這會導致記憶體的 Call Stack 毫無節制地瘋狂擴張，陷入邏輯失控的自我複製，最終以系統崩潰慘烈收場。這完美呼應了講義所強調的：**必須遇到 Base Case，大事化小的任務才能輕鬆解決並開始 Go Up（往上層回溯）。**

##七、 演算法思維批判：分治法 vs 枚舉法的本質差異
根據講義前段（第 4~5 頁）指出，**枚舉法 (Enumeration)** 旨在將所有可能的路線與解答全部走過一遍，是一種相對盲目且低效的暴力解策略。而分治法則是精妙地「大事化小，小事成大」。
####深度批判
然而，在河內塔這個極端的案例中，**分治法在時間複雜度上並沒有真正減少任何一絲一毫的絕對運算量。**因為河內塔的物理限制极其強烈（小盤絕不可壓在大盤之上），要移開下方的結構，上方的所有盤子就必須老老實實地移開，其最少移動次數在數學物理本質上就是固定的 $2^N - 1$。換句話說，此處的分治法實質上是「一種具有高度秩序的暴力枚舉」。它優化的是「人類理解問題的架構、程式撰寫的邏輯深度與表達方式」，而不是「絕對運算的物理次數」。它把原本盲目的暴力枚舉，整理成了極具幾何結構規律的遞迴流程。

##八、 工程語義分析：大規模計算的致命傷與啟示
在 $N = 1,000,000$ 時，該工程問題存在著兩個層次的不可行性：


1. **運算次數的組合爆炸：** 總移動次數為 $2^{1,000,000}-1$，這是一個超越宇宙原子總數的通天數字，在物理時間尺度上完全不可行。
2. **記憶體資源的耗盡：** 縱使不執行實質移動，單純的遞迴切分所產生的 100 萬層系統 Call Stack 棧幀，也會在瞬間吃光執行緒的棧區記憶體，造成致命的 Stack Overflow。

###工程實務啟示
這對工程師帶來了關鍵的警示：在面對大規模的工程數據時（例如：**超大規模土木結構點雲資料處理、高階非線性結構動態分析矩陣、大範圍地理資訊系統 GIS 網格計算**），我們絕不能盲目地追求程式碼的簡潔與邏輯優雅而濫用深層遞迴。必須隨時評估硬體的邊界限制，適時引入 **Iterative Stack、Queue 或是記憶體可控的動態規劃（Dynamic Programming）** 方法，才能確保工程系統的穩定性與強健性。
